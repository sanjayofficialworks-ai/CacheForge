```python
queue = {'task1': 'login', 'task2': 'load', 'task3': 'display'}

# Grab the very first key in the insertion order
oldest_key = next(iter(queue))

# Remove and return that specific key's value
oldest_value = queue.pop(oldest_key)

print(f"Removed: {oldest_key} -> {oldest_value}")
# Output: Removed: task1 -> login
```
